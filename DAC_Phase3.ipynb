```python
import pandas as pd 
import numpy as np 
import seaborn as sns
import matplotlib.pyplot as plt 
print('Modules are imported.')
df=pd.read_csv("/Users/kisshore/Downloads/Covid_19_cases4.csv")
print(df.head())
print(df.shape)
df.drop(["dateRep","day","month","year"],axis=1,inplace=True)
print(df.head())
aggregating=df.groupby("countriesAndTerritories").sum()
print(aggregating.head())
print(aggregating.shape)
aggregating.loc["Austria"].plot()
aggregating.loc["Finland"].plot()
aggregating.loc["Cyprus"].plot()
plt.legend()
plt.show()
```
